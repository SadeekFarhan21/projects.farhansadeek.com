```python
import pandas as pd
import matplotlib.pyplot as plt

# Create or load your DataFrame
# For example:
data = {
    'chapter': ['A', 'A', 'B', 'B', 'C', 'C', 'C'],
    'lrn_type': ['Type1', 'Type2', 'Type1', 'Type2', 'Type1', 'Type2', 'Type1'],
    'points_possible': [10, 10, 15, 15, 20, 20, 20],
    'points_earned': [8, 9, 12, 13, 18, 17, 19]
}
df = pd.DataFrame(data)

# Group by 'chapter' and 'lrn_type', then count the occurrences of 'lrn_type'
temp = responses.groupby(['chapter_number', 'lrn_type']).size().unstack(fill_value=0)

# Plot the results
temp.plot(kind='bar', stacked=True)
plt.xlabel('Chapter')
plt.ylabel('Number of Questions')
plt.title('Number of Questions by Learning Type for Each Chapter')
plt.xticks(rotation=0)
plt.legend(title='Learning Type')
plt.show()
```

```python
responses['accuracy'] = responses['points_earned'] / responses['points_possible']
accuracy_by_type = responses.groupby('lrn_type')['accuracy'].mean()
accuracy_by_type
```
